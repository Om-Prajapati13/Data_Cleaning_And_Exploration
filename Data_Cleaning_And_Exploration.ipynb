```python
# Load Dataset
import pandas as pd
df = pd.read_csv('Titanic-Dataset.csv')
print("Dataset Loaded")
```

```python
# Clean Data
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop(columns=['Cabin'])
df = df.drop_duplicates()
df['Survived'] = df['Survived'].astype('category')
df['Pclass'] = df['Pclass'].astype('category')
print("Missing values handled, duplicates removed, and data types corrected")
```

```python
# Display Basic Statistics
print("Mean Age:", df['Age'].mean())
print("Median Age:", df['Age'].median())
print("Survival Value Counts:")
print(df['Survived'].value_counts())
print("Passenger Class Value Counts:")
print(df['Pclass'].value_counts())
```

```python
# Summarize Cleaning steps
report = """
Summary of Cleaning Steps:
1. Handled missing 'Age' values by filling with the median.
2. Handled missing 'Embarked' values by filling with the mode.
3. Dropped the 'Cabin' column due to excessive missing values.
4. Removed duplicate rows.
5. Converted 'Survived' and 'Pclass' columns to categorical data types.
"""
print(report)
```
